```python
import pandas as pd #별칭

# 1차원 리스트 데이터 생성 - Series 클래스
data = ['식빵', '딸기잼', '와인']
s1 = pd.Series(data, index=[1, 2, 3])
print(s1)
print(s1.index) #원래는 0부터 시작
print(s1.values) #값: ['식빵', '딸기잼', '와인']
print(s1.shape) #(3,) 열이 없고 행만 있다
```

데이터프레임(DataFrame) - 행과 열로 이루어진 2차원 데이터 구조이다

```python
import pandas as pd

# 데이터프레임 생성
df = pd.DataFrame({
   "name":['키보드', '마우스', 'USB 메모리'],
   "price":[20000, 30000, 11000] 
})
print(df)
print(df.columns) #Index(['name', 'price'], dtype='str')
print(df.shape) #(3, 2) 3행 2열

# 검색
print(df['name']) #검색 대괄호
print(df[['name', 'price']]) #2개 열이므로, 리스트를 [] 한번더 감싸주고 검색[] 해야한다

# 칼럼 이름을 한글로 작서
# df.rename(columns={"name" : "품명", "price" : "가격"}, inplace=True) #inplace=True 즉시저장
df = df.rename(columns={"name" : "품명", "price" : "가격"})
print(df)
```

```python
# 칼럼 추가
df['사양(spec)'] = ["유선", "무선", "128GB"]
print(df)
```
