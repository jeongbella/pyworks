```python
import pandas as pd

food = pd.DataFrame({
    "메뉴": ["김밥", "비빔밥", "자장면", "우동", "김밥", "자장면", "김밥"],
    "가격": [3000, 7000, 5500, 5500, 3500, 6000, 4000],
    "분류": ["한식", "한식", "중식", "일식", "한식", "중식", "한식"]
})
# print(food)

# 데이터의 일부만 검색 - head(), tail()
# print(food.head())
# print(food.tail())

# 데이터 정보 - info()
food.info()

# 요약 정보 - describe() 
food.describe() #기본-수치형 칼럼 요약 (디폴트는 숫자출력)
food.describe(include='object') #문자형 칼럼 요약
food.describe(include='number') #수치형 칼럼 요약

# 고유한 값의 개수 (중복제외)
food["메뉴"].value_counts()
food["분류"].value_counts()
```

```python
#  조건 검색 (where 절이라고 생각)
# 가격이 4000원 미만인 음식의 정보 
result1 = (food["가격"] < 4000)
print(food[result1])

# 메뉴가 김밥이 아닌 음식 정보 (아니다 !=)
print(food[food["메뉴"] != "김밥"])

# 가격을 내림차순으로 정렬 - sort_values()
print(food.sort_values("가격", ascending=False))
```

```python
# 파이썬 : sum(리스트), len(리스트)
# 집계(통계) 함수 
food["판매개수"] = [30, 70, 55, 55, 30, 60, 40]
# print(food)
print(food["판매개수"].count()) #개수
print(food["판매개수"].sum()) #합계
print(round(food["판매개수"].mean(), 2)) #평균(소수 둘째자리)

# 최빈값 - 자주 등장하는 값 : mode()
print(food["메뉴"].mode()) #시리즈 형
print(food["메뉴"].mode()[0]) #김밥

# 그룹핑 - groupby()
# 분류별 판매개수의 합계
print(food.groupby("분류")["판매개수"].sum())
# 메뉴별 판매개수의 최대값 계산
print(food.groupby("분류")["판매개수"].max())
```

```python
# food 데이터프레임 만들기
food.to_csv("datas/food.csv", index=False)
```
